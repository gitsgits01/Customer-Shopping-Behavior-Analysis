```python
x= df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns = {'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
#creating  new column age_group
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_groups'] = pd.qcut(df['age'], q=4, labels= labels)
```

```python
df[['age','age_groups']].head(10)
```

```python
#Creating purchase_frequency _days
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly' : 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```
